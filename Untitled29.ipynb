```python

import random

# User input
p = float(input("Enter perimeter of garden: "))
n = int(input("Enter number of cells: "))
it = int(input("Enter number of iterations: "))

# Initialize cells
cells = [random.uniform(0, p/2) for i in range(n)]

def fitness(x):
    return x * (p/2 - x)

# Parallel Cellular Algorithm
for k in range(it):
    old = cells.copy()
    new = []

    for i in range(n):
        neighbors = [
            old[(i-1) % n],
            old[i],
            old[(i+1) % n]
        ]

        best = max(neighbors, key=fitness)

        x = best + random.uniform(-p/20, p/20)
        x = max(0, min(p/2, x))

        new.append(x)

    cells = new

# Best solution
best = max(cells, key=fitness)
width = p/2 - best
area = fitness(best)

print("\nBest Length:", round(best, 2))
print("Best Width:", round(width, 2))
print("Maximum Area:", round(area, 2))
```
