```python

import random

# User input
a = float(input("Enter cost factor (a): "))
b = float(input("Enter revenue factor (b): "))
c = float(input("Enter fixed cost (c): "))
limit = int(input("Enter maximum production quantity: "))
n = int(input("Enter population size: "))
gen = int(input("Enter number of generations: "))
mr = float(input("Enter mutation rate: "))
cr = float(input("Enter crossover rate: "))

bits = 12

# Gene expression: Binary to production quantity
def express(ch):
    value = int(ch, 2)
    return round(value * limit / (2**bits - 1))

# Fitness function
def fitness(ch):
    q = express(ch)
    return -a*q*q + b*q - c

# Initialize population
pop = [
    ''.join(random.choice('01') for _ in range(bits))
    for i in range(n)
]

# Gene Expression Algorithm
for g in range(gen):

    new = []

    # Elitism: Keep the best chromosome
    best = max(pop, key=fitness)
    new.append(best)

    while len(new) < n:

        # Selection
        p1 = max(random.sample(pop, 3), key=fitness)
        p2 = max(random.sample(pop, 3), key=fitness)

        # Crossover
        if random.random() < cr:
            point = random.randint(1, bits-1)
            c1 = p1[:point] + p2[point:]
            c2 = p2[:point] + p1[point:]
        else:
            c1, c2 = p1, p2

        # Mutation
        children = [c1, c2]

        for ch in children:
            ch = list(ch)

            for i in range(bits):
                if random.random() < mr:
                    ch[i] = '1' if ch[i] == '0' else '0'

            new.append(''.join(ch))

            if len(new) == n:
                break

    pop = new

# Best solution
best = max(pop, key=fitness)

print("\nBest Chromosome:", best)
print("Optimal Production Quantity:", express(best))
print("Maximum Profit:", round(fitness(best), 2))
```
