```python
import numpy as np

# --- 1. PROBLEM SETUP & CONSTRAINTS ---
# Hourly electricity rates ($ per kWh) for a 24-hour day
# Cheap at night (0-6), expensive in afternoon/evening (16-21)
HOURLY_RATES = np.array([
    0.08, 0.08, 0.08, 0.07, 0.07, 0.08,  # 00:00 - 05:00
    0.12, 0.15, 0.15, 0.12, 0.10, 0.10,  # 06:00 - 11:00
    0.15, 0.22, 0.35, 0.40, 0.45, 0.42,  # 12:00 - 17:00
    0.38, 0.30, 0.20, 0.15, 0.10, 0.09   # 18:00 - 23:00
])

BATTERY_CAPACITY_KWH = 50.0  # Total capacity to add
MAX_CHARGING_POWER_KW = 10.0  # Max power the charger can pull per hour
EFFICIENCY = 0.90             # 90% round-trip charging efficiency

# --- 2. GENETIC ALGORITHM PARAMETERS ---
POP_SIZE = 100         # Number of schedules in each generation
GENERATIONS = 200      # How many iterations to run
MUTATION_RATE = 0.15   # Probability of a random schedule change
ELITISM_COUNT = 5      # Keep the top X best schedules automatically

# --- 3. CORE FUNCTIONS ---

def calculate_fitness(schedule):
    """
    Evaluates a schedule. Returns (cost, penalty).
    We want to MINIMIZE both cost and capacity deficits.
    """
    # Actual energy added to the battery per hour factoring in efficiency
    energy_added_per_hour = schedule * MAX_CHARGING_POWER_KW * EFFICIENCY
    total_energy_charged = np.sum(energy_added_per_hour)

    # Financial cost calculation
    cost = np.sum(schedule * MAX_CHARGING_POWER_KW * HOURLY_RATES)

    # Heavy penalty if the schedule fails to hit the target capacity
    capacity_deficit = max(0.0, BATTERY_CAPACITY_KWH - total_energy_charged)
    penalty = capacity_deficit * 1000.0  # Severe multiplier for failing constraints

    # Total score (lower is better)
    return cost + penalty, cost, total_energy_charged

def initialize_population():
    """Generates random schedules where each hour is a value between 0.0 and 1.0."""
    return np.random.uniform(0.0, 1.0, size=(POP_SIZE, 24))

def selection(pop, scores):
    """Selects a parent using Tournament Selection."""
    tournament_idx = np.random.choice(len(pop), size=3, replace=False)
    best_idx = tournament_idx[np.argmin(scores[tournament_idx])]
    return pop[best_idx]

def crossover(parent1, parent2):
    """Combines two schedules using a single-point split."""
    split = np.random.randint(1, 23)
    child1 = np.concatenate([parent1[:split], parent2[split:]])
    child2 = np.concatenate([parent2[:split], parent1[split:]])
    return child1, child2

def mutate(schedule):
    """Randomly alters the charging rate of specific hours to preserve diversity."""
    for hour in range(24):
        if np.random.rand() < MUTATION_RATE:
            # Shift the charging rate up or down randomly, keeping it between 0 and 1
            schedule[hour] = np.clip(schedule[hour] + np.random.uniform(-0.3, 0.3), 0.0, 1.0)
    return schedule

# --- 4. EXECUTION LOOP ---

def run_optimization():
    population = initialize_population()

    print(f"Starting Genetic Algorithm Optimization over {GENERATIONS} generations...")

    for gen in range(GENERATIONS):
        # 1. Evaluate current generation
        evaluations = [calculate_fitness(ind) for ind in population]
        fitness_scores = np.array([e[0] for e in evaluations])
        costs = np.array([e[1] for e in evaluations])
        capacities = np.array([e[2] for e in evaluations])

        # Track the absolute best item from this generation
        best_idx = np.argmin(fitness_scores)

        if gen % 40 == 0 or gen == GENERATIONS - 1:
            print(f"Gen {gen:3d} | Best Strategy Cost: ${costs[best_idx]:.2f} | Charged: {capacities[best_idx]:.1f}/{BATTERY_CAPACITY_KWH} kWh")

        # 2. Build the next generation
        next_population = []

        # Elitism: Keep the best performers unchanged
        sorted_indices = np.argsort(fitness_scores)
        for i in range(ELITISM_COUNT):
            next_population.append(population[sorted_indices[i]])

        # Breeding loop to fill the rest of the population
        while len(next_population) < POP_SIZE:
            p1 = selection(population, fitness_scores)
            p2 = selection(population, fitness_scores)

            c1, c2 = crossover(p1, p2)

            next_population.append(mutate(c1))
            if len(next_population) < POP_SIZE:
                next_population.append(mutate(c2))

        population = np.array(next_population)

    # Return the optimal layout found at the end
    final_evals = [calculate_fitness(ind) for ind in population]
    final_best_idx = np.argmin([f[0] for f in final_evals])
    return population[final_best_idx], final_evals[final_best_idx]

# --- 5. REPORT RESULTS ---
if __name__ == "__main__":
    best_schedule, (total_score, final_cost, final_capacity) = run_optimization()

    print("\n" + "="*40)
    print("OPTIMIZED 24-HOUR CHARGING SCHEDULE")
    print("="*40)
    print(f"Total Charging Cost: ${final_cost:.2f}")
    print(f"Total Energy Delivered: {final_capacity:.2f} kWh (Target: {BATTERY_CAPACITY_KWH} kWh)")
    print("-"*40)
    print("Hour  | Electricity Rate ($/kWh) | Charger Utilization")
    print("-"*40)
    for hr, utilization in enumerate(best_schedule):
        print(f"{hr:02d}:00 | ${HOURLY_RATES[hr]:.2f}                   | {utilization * 100:5.1f}%")
```
